# Environmental Footprints of Housing
This code calculates environmental footprints of housing under a given set of energy use, energy technology, housing size of different building types, and transport use in relation to building type, cohort, location and income or tenure.

For the analysis of current housing in NL the following datasets are needed:
- Set of assumptions: CBS microdata output
- Material use per building types and cohorts
- Emission factors per pkm, kg, kWh, solar PV capacity, heat pump

```python
# data processing modules
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer # used for imputation of missing mobility behaviour data

# data visualisation modules
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.gridspec import GridSpec
import matplotlib.patches as mpatches

# Display options
pd.options.display.float_format = '{:,.2f}'.format #limits printed decimal points to two
#pd.reset_option('^display.', silent=True) #option to reset the previous display option
```

```python
def bubblechart(total, per, bubblesize, y_level, category_level, xlabel, ylabel, filename, bubblecolors=['#ffb000', '#fe6100','#dc267f','#785ef0', '#648fff','#808080']): 
    #floor area per income level
    use_m2_total_class = total.groupby(level=y_level, sort=False).sum() #m2 total in each group summed by income level
    use_cap_total_class = per.groupby(level=y_level, sort=False).sum() #no. of people in each group summed by income level
    use_m2_per_cap_class = use_m2_total_class / use_cap_total_class #m2/cap in each income group

    #floor area per income level and building type
    use_m2_total_class_type = total.groupby(level=[y_level,category_level], sort=False).sum() #m2 total in each group summed by income level
    use_cap_total_class_type = per.groupby(level=[y_level,category_level], sort=False).sum() #no. of people in each group summed by income level
    use_m2_per_cap_class_type = use_m2_total_class_type / use_cap_total_class_type #m2/cap in each income group

    #preparing the dataframes for plotting
    df = use_m2_per_cap_class
    df2=use_m2_per_cap_class_type.rename('value').to_frame()
    df2['n_hhd'] = bubblesize.groupby(level=[y_level,category_level], sort=False).sum()
    df2 = df2.reset_index()

    #plot
    colors = ['#000000']#,'#fe6100','#648fff', '#785ef0','#ffb000','#808080','#000000', ]
    fig = px.bar(df, x=df.values, y=df.index,  
                labels={'x':xlabel, lab:ylabel, 'density_lab':ylabel}, 
                template='simple_white', 
                color_discrete_sequence=colors) 
    fig.update_layout(width=1000, height=500, 
                        font=dict(size=18),
                        bargap=0.9,
                        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="left",
            x=-0.25,
        ))

    df2=use_m2_per_cap_class_type.rename('value').to_frame()
    df2['n_hhd'] = bubblesize.groupby(level=[y_level,category_level], sort=False).sum()
    
    count=0
    for i in df2.index.get_level_values(1).unique().to_list():
        df_plot = df2.xs(i, level=1)
        fig.add_trace(go.Scatter(
                y=df_plot.index, 
                x=df_plot['value'],
                mode='markers',
                name=i,
                legendrank=count,
                marker=dict(size=df_plot['n_hhd'],
                            sizemode='area',
                            sizeref=4*max(df2['n_hhd'])/(40.**2),
                            sizemin=4,
                            color=bubblecolors[count],
                            ),
                showlegend=True
            ))
        count=count+1

    fig.write_image('visualisations/'+class_name_short+'/bubblechart_'+filename+year+age_group+inc_grouping+'.png', scale=3)
    fig.show()
```

```python
# Configurations
#scenario = 'current'
scenario = 'Boekel'
year = '2023' #alternative: '2019', '2023'
age_group = ''  #alternative: '' means all ages of household heads; '_age1864' means household heads are aged 18-64 i.e. working age population, '_incclass' for income levels in four classes including top1%
classification = 'income' #alternative: 'tenure', 'income'
if classification == 'income':
    lab = 'inc_lab'
    class_name = 'income level'
    class_name_short = 'income'
    class_name_long = 'income level'
    if age_group == '_incclass':
        inc_grouping = 'top1deciles' #alternative: '' predefined groups or deciles, 'top1deciles' replaces decile 10 (richest) by top 1% and 90-99%
        class_col = 'inc_decile' #'inc_class'+year
    else:
        inc_grouping = ''
        class_col = 'inc_decile'
else: #if tenure
    lab = 'tenure_lab'
    class_name = 'tenure'
    class_name_short = 'tenure'
    class_name_long = 'tenure'
    class_col = str('tenure'+year)
    inc_grouping = ''
material_buildings_source = 'Oorschot-MPG'
ghg_pkm_source = 'Milieucentraal' #alternative: 'Milieucentraal' assumes average cars in NL, 'Milieucentraal - e-car' assumes all cars to be e-cars
ghg_districtheating_source = 'BONSAI' #alternative: 'BONSAI' assumes roughly 1.5 the emission intensity of 'Verhagen', 'Herberigs' assumes 1.4 times emission intensity of Verhagen

# File import
#if scenario == "current":
if scenario == "Boekel":
    if inc_grouping == 'top1deciles':
        use = pd.read_excel(
                    "data_input/output/"+class_name_short+" typology construction density distance privacy "+year+".xlsx",
                    index_col=[0,1,2,3,4],
                    header=[0],
                    )
        use.drop('decile 10 (richest)',axis=0, level=0, inplace=True)
                
        use_top1 = pd.read_excel(
                    "data_input/output/"+class_name_short+" typology construction density distance privacy "+year+age_group+".xlsx",
                    index_col=[0,1,2,3,4],
                    header=[0],
                    )
        use_top1.rename(columns={'inc_class'+year: 'inc_decile'}, inplace=True)
        use_top1.drop(['51-90%', 'poorest 50%'],axis=0, level=0, inplace=True)
        use_top1.rename(index={'91-99%':'percentile 91-99%', 'top 1%':'top 1% (richest)'}, inplace=True)

        use_top1.replace({'inc_decile': 4}, {'inc_decile': 11}, inplace=True)
        use_top1.replace({'inc_decile': 3}, {'inc_decile': 10}, inplace=True)
        use = pd.concat([use, use_top1])

        mobility = pd.read_excel(
                        "data_input/output/"+class_name_short+" density distance privacy odin "+year+".xlsx",
                        index_col=[0,1,2],
                        header=[0],
                        )
        mobility.drop('decile 10 (richest)',axis=0, level=0, inplace=True)
                
        mobility_top1 = pd.read_excel(
                        "data_input/output/"+class_name_short+" density distance privacy odin "+year+age_group+".xlsx",
                        index_col=[0,1,2],
                        header=[0],
                        )
        mobility_top1.rename(columns={'inc_class'+year: 'inc_decile'}, inplace=True)
        mobility_top1.drop(['51-90%', 'poorest 50%'],axis=0, level=0, inplace=True)
        mobility_top1.rename(index={'91-99%':'percentile 91-99%', 'top 1%':'top 1% (richest)'}, inplace=True)

        mobility_top1.replace({'inc_decile': 4}, {'inc_decile': 11}, inplace=True)
        mobility_top1.replace({'inc_decile': 3}, {'inc_decile': 10}, inplace=True)

        mobility = pd.concat([mobility, mobility_top1])
    else:
        use = pd.read_excel(
            "data_input/output/"+class_name_short+" typology construction density distance privacy "+year+age_group+".xlsx",
            index_col=[0,1,2,3,4],
            header=[0],
        )

        mobility = pd.read_excel(
                "data_input/output/"+class_name_short+" density distance privacy odin "+year+age_group+".xlsx",
                index_col=[0,1,2],
                header=[0],
            )
    
    #fill missing information for small sample observations with average value
    mobility.loc[mobility['n_hhd_uw'].notna()] = mobility.loc[mobility['n_hhd_uw'].notna()].fillna(5)

    # add transport use and vehicles per household to other variables, assumption: households with similar income, surrounding population density and distance to train station show similar mobility behaviour irrespective of building type or building age
    use_pkm = mobility['dist_carpp_py'] / mobility['n_hhd']#pkm per year per household
    use_pkm.name = 'car_pkm'
    use = use.join(use_pkm, on=[lab,'density_lab', 'distance_train_lab'])

    use_vkm = mobility['dist_car_py'] / mobility['n_hhd']#pkm per year per household
    use_vkm.name = 'car_vkm'
    use = use.join(use_vkm, on=[lab,'density_lab', 'distance_train_lab'])

    use_pkm = mobility['dist_transit_py'] / mobility['n_hhd'] #pkm per year per household
    use_pkm.name = 'transit_pkm'
    use = use.join(use_pkm, on=[lab,'density_lab', 'distance_train_lab'])

    use_pkm = mobility['dist_walk_py'] / mobility['n_hhd'] #pkm per year per household
    use_pkm.name = 'walk_pkm'
    use = use.join(use_pkm, on=[lab,'density_lab', 'distance_train_lab'])

    use_pkm = mobility['dist_cycle_py'] / mobility['n_hhd'] #pkm per year per household
    use_pkm.name = 'cycle_pkm'
    use = use.join(use_pkm, on=[lab,'density_lab', 'distance_train_lab'])

    use_pkm = mobility['car_hhd'] / mobility['n_hhd_uw'] #cars per household
    use_pkm.name = 'car_hhd'
    use = use.join(use_pkm, on=[lab,'density_lab', 'distance_train_lab'])

    #weighting the pkm depending on number of adults in household i.e. more pkm allocated to households with more adults, less to households with fewer adults
    use_cap_per_hhd_mean = (use['n_adultmem'+year].groupby(level=[0,3,4], sort=False).sum()/use['n_hhd'].groupby(level=[0,3,4], sort=False).sum()) 
    use_cap_per_hhd_mean.name = 'n_mem_per_hhd_mobility'
    use = use.join(use_cap_per_hhd_mean, on=[lab,'density_lab', 'distance_train_lab'])
    use['mem_weight'] = (use['n_adultmem'+year]/use['n_hhd'])/use['n_mem_per_hhd_mobility']

    for i in ['car_pkm', 'car_vkm', 'transit_pkm', 'walk_pkm', 'cycle_pkm','car_hhd']:
        use[i] = use[i]*use['mem_weight']
    
    #removing entries where no observations in use dataset
    use.loc[use['size'+year+'_w1'].isna(), 'car_vkm'] = np.NaN
    use.loc[use['size'+year+'_w1'].isna(), 'n_mem_per_hhd_mobility'] = np.NaN

    #impute entries where no observations of mobility behaviour (low income+rural+close to train station) roughly 0.1% of population
    index_cols = [class_col, 'typology4_'+year, 'const_cat'+year, 'density'+year,
       'q_distance_train'+year]
    mobility_cols = ['car_pkm', 'car_vkm', 'transit_pkm', 'walk_pkm',
        'cycle_pkm', 'car_hhd']
    pred_cols = ['distance_depstore'+year, 'distance_supermarket'+year, 'distance_middleschool'+year, 'mean_hhd', 'mean_adult']
    df = use.loc[use['n_hhd'].notna(), index_cols+pred_cols+mobility_cols]
    use_imputed = pd.DataFrame(KNNImputer(missing_values=np.nan, n_neighbors=3, weights='distance').fit(df).transform(df), index=df.index, columns=df.columns)
    use.update(use_imputed)

    #change index labels to create more readable graphs
    use = use.rename(index={'detached (vrijstaand)':'detached', 'multiple fam (meergezins)':'multifamily', 'row house (tussenwoning)':'row house', 'semi-detached (2-1 kap) / corner (hoek)':'semi-detached'}, level=1)
    use = use.rename(index={'priate rental':'private rental'}, level=0)

    
    # output the combined file (including building attributes and mobility behaviour)
    use.to_csv('data_input/output/use'+class_name_short+year+age_group+inc_grouping+'.csv')
```

```python
# Loading the material intensities per floor area
if material_buildings_source == "Oorschot-MPG": #kg/m2
    # --- BOEKEL OVERRIDE: fixed materials and intensities (skip Excel) ---
    if scenario == "Boekel":
        # hard-coded total masses per dwelling [kg/dwelling]
        boekel_materials = {
            "Glass foam": 21470.0,
            "Geopolymer concrete": 111114.25,
            "Screed floors": 83500.0,
            "Lime hemp walls": 42937.5,
            "Wood (timber)": 19910.42,
        }
        # hard-coded GHG intensities [kgCO2e/kg]
        boekel_intensities = {
            "Glass foam": 1.34,
            "Geopolymer concrete": 0.092,
            "Screed floors": 0.11,
            "Lime hemp walls": -0.798,
            "Wood (timber)": 0.30,
        }

        # Store them as DataFrames for later steps
        mi_buildings = pd.DataFrame(
            [boekel_materials],  # kg/dwelling
            index=pd.MultiIndex.from_tuples(
                [("Boekel_typology", "Boekel_construction")],
                names=["typology_lab", "const_lab"],
            ),
        )
        ghg_intensity_buildings = pd.DataFrame(
            [boekel_intensities], index=mi_buildings.index
        )[mi_buildings.columns]

    else:
        mi_buildings = pd.read_excel(
            "data_input/MI_Buildings_Oorschot2023MetabolicMPG.xlsx",
            sheet_name="Building_MI",
            index_col=[0, 1],
            header=[0, 1], skiprows=3
        ).drop("2018-2050", axis=0, level=1)
        mi_buildings.index.names = ["Building type", "Cohort"]
        mi_buildings.columns.names = ["BONSAI", "Material"]
        mi_buildings.fillna(0, inplace=True)
```

```python
# --- Calculating the material contained in an average dwelling per category ---
if scenario == "Boekel":
    #fixed_floor_area = 78  # m²/dwelling

    # Keep EXACTLY the same index as `use` and just set size to 78 everywhere
    use_m2 = use['size' + year + '_w1'].copy()
    use_m2.loc[:] = fixed_floor_area  # preserve index names & labels

    # Broadcast the single-row mi_buildings (kg/dwelling) to all rows of `use`
    # (assumes mi_buildings has one row with your hard-coded materials)
    base = mi_buildings.iloc[0]
    material_buildings = pd.DataFrame(
        [base.values] * len(use.index),
        index=use.index,                    # <-- keep original index exactly
        columns=mi_buildings.columns
    )

else:
    # Original path for non-Boekel (no renaming of index levels)
    use_m2 = use['size' + year + '_w1']  # m²/dwelling

    # Replicate m² across material columns
    use_m2 = pd.concat([use_m2] * len(mi_buildings.columns), axis=1)
    use_m2.columns = mi_buildings.columns

    # Multiply m² by kg/m² → kg/dwelling; match on the existing first N index levels of mi_buildings
    material_buildings = use_m2.multiply(
        mi_buildings.reindex(use_m2.index),
        level=list(range(mi_buildings.index.nlevels))
    )

# Result
material_buildings  # kg / dwelling, index unchanged

```

```python
# --- calculating embodied GHG emissions of materials in buildings ---
if scenario == "Boekel":
    # Ensure we have a 1D intensity Series aligned to material names
    if isinstance(ghg_intensity_buildings, pd.DataFrame):
        ghg_intensity_series = ghg_intensity_buildings.iloc[0]
    else:
        ghg_intensity_series = ghg_intensity_buildings

    # Multiply material masses by intensities, handling single- or multi-level columns
    if isinstance(material_buildings.columns, pd.MultiIndex):
        # Map intensities by the last level (material names)
        last_level = material_buildings.columns.get_level_values(-1)
        factors = last_level.map(ghg_intensity_series).astype(float).to_numpy()
        ghg_material_buildings = material_buildings.multiply(factors, axis=1)
    else:
        # Single-level columns: align by column name
        factors = ghg_intensity_series.reindex(material_buildings.columns).astype(float)
        ghg_material_buildings = material_buildings.multiply(factors, axis=1)

else:
    # --- original BONSAI-based path ---
    mult_bonsai = pd.read_csv(
        'data_input/240605_BONSAI_ghg_for_all_iluc.csv',
        index_col=[0,1,2,3], header=[0], skipfooter=9, engine="python"
    )  # kg CO2eq per kg material

    # selecting relevant product codes (first level of columns)
    prod_codes = material_buildings.columns.levels[0].to_list()
    prod_codes.append('M_CMNT')
    mult_bonsai = mult_bonsai[mult_bonsai.index.get_level_values(0).isin(prod_codes)]
    mult_bonsai = mult_bonsai.droplevel([1,2,3])
    mult_bonsai = mult_bonsai[['NL']]  # NL market

    # composite concrete from cement + sand/gravel mix (example)
    concrete_cement_pc = 0.35  # M20–M25
    mult_bonsai.loc['M_CMNT + M_SDCL'] = (
        mult_bonsai.loc[['M_CMNT']].squeeze()*concrete_cement_pc
        + mult_bonsai.loc[['M_SDCL']].squeeze()*(1-concrete_cement_pc)
    )

    # group materials to product-code level and multiply by intensities
    material_buildings_grouped = material_buildings.groupby(axis=1, level=0).sum(min_count=1)  # kg/dwelling
    ghg_material_buildings = material_buildings_grouped.multiply(mult_bonsai.iloc[:,0], axis=1).dropna(axis=1, how='all')

# final object:
ghg_material_buildings  # kg CO2e/dwelling
```

### Building Use Phase

#### Electricity

```python
# --- Electricity GHG (building use) ---
mult_bonsai = pd.read_csv(
        'data_input/240605_BONSAI_ghg_for_all_iluc.csv',
        index_col=[0,1,2,3], header=[0], skipfooter=9, engine="python"
    )  # kg CO2eq per kg material
mult_bonsai = mult_bonsai['NL']  # select only NL market

if scenario == "Boekel":
    # Negative electricity balance in Boekel scenario (solar feed-in to grid)
    electricity_buildinguse = use['netelectricity' + year].copy()
    electricity_buildinguse.loc[:] = 0.0  # kWh/household/yr (not used for GHG)
    
    # offset
    ghg_electricity_buildinguse = electricity_buildinguse.copy()
    ghg_electricity_buildinguse.loc[:] = 0  # kg CO2e/household/yr

else:
    # Import environmental footprint multipliers
    mult_bonsai = pd.read_csv(
        'data_input/240605_BONSAI_ghg_for_all_iluc.csv',
        index_col=[0,1,2,3],
        header=[0],
        skipfooter=9,
        engine="python"
    ).droplevel([1,2,3])  # tonne CO2eq per TJ electricity

    # kWh -> MJ -> TJ per household per year
    electricity_buildinguse = use['netelectricity' + year]   # kWh/hh/yr
    electricity_buildinguse = electricity_buildinguse * 3.6  # MJ/hh/yr
    electricity_buildinguse = electricity_buildinguse / 1e6  # TJ/hh/yr

    # tonne CO2e/TJ * TJ -> tonne; *1000 -> kg CO2e/hh/yr
    ghg_electricity_buildinguse = electricity_buildinguse * mult_bonsai.loc['M_electricity'] * 1000

# Result
ghg_electricity_buildinguse  # annual GHG emission per household (kg CO2e/yr)
```

#### Gas

```python
# --- Natural gas GHG (building use) ---
if scenario == "Boekel":
    # No gas consumption in Boekel scenario
    gas_buildinguse = use['gas' + year].copy()
    gas_buildinguse.loc[:] = 0.0  # m³/household/year

    # All derived quantities remain zero
    gas_buildinguse_tonne = gas_buildinguse.copy()
    gas_buildinguse_GJ = gas_buildinguse.copy()
    ghg_gas_buildinguse_footprint = gas_buildinguse.copy()
    ghg_gas_buildinguse_direct = gas_buildinguse.copy()
    ghg_gas_buildinguse = gas_buildinguse.copy()

else:
    # M_GASE: upstream emissions per tonne
    # EF_GASE: combustion emissions per TJ
    gas_buildinguse = use['gas' + year]  # m³ per household per year

    # Convert to tonnes and GJ
    gas_buildinguse_tonne = gas_buildinguse * 0.735  # kg per household per year
    gas_buildinguse_GJ = gas_buildinguse * 0.0316    # GJ per household per year

    # Apply emission factors
    ghg_gas_buildinguse_footprint = gas_buildinguse_tonne * mult_bonsai.loc['M_GASE']  # kg CO2eq/hh/yr (upstream)
    ghg_gas_buildinguse_direct = gas_buildinguse_GJ * mult_bonsai.loc['EF_GASE']       # kg CO2eq/hh/yr (combustion)

    # Total GHGs from natural gas
    ghg_gas_buildinguse = ghg_gas_buildinguse_footprint + ghg_gas_buildinguse_direct

# Result
ghg_gas_buildinguse  # annual GHG emission per household (kg CO2eq/yr)
```

#### Heat pump

```python
# --- Heat pump GHG (building use) ---
if scenario == "Boekel":
    # No heat pumps in Boekel scenario
    heatpump_buildings = use['n_heatpump' + year].copy()
    heatpump_buildings.loc[:] = 0.0  # mean number per household
    ghg_heatpump_buildings = heatpump_buildings.copy()  # 0 kg CO2e/household

else:
    # Calculate mean number of heat pumps per household
    heatpump_buildings = use['n_heatpump' + year] / use['n_hhd']

    # Apply emission factor (Verhagen et al., 2021)
    ghg_heatpump_buildings = heatpump_buildings * 386  # kg CO2e per household

# Result
ghg_heatpump_buildings  # total GHG emission per household (kg CO2e/yr)
```

#### District heating

```python
# --- District heating GHG (building use) ---
if scenario == "Boekel":
    # No district heating use in Boekel scenario
    energy_districtheating = use['n_cityheating' + year].copy()
    energy_districtheating.loc[:] = 0.0  # GJ per household per year
    ghg_heat_districtheating = energy_districtheating.copy()  # 0 kg CO2e/household/year

else:
    # Calculate GJ heat delivered per household per year
    energy_districtheating = (
        use['n_cityheating' + year] * use['warmteschatting_SW'] / use['n_hhd']
    )  # GJ heat delivered/household/yr

    # Account for pre-distribution generation (24% losses)
    energy_districtheating = energy_districtheating * 1.32  # GJ generated pre-distribution

    # Apply emission factors depending on data source
    if ghg_districtheating_source == 'BONSAI':
        # BONSAI: roughly 1.5× Verhagen
        ghg_heat_districtheating = energy_districtheating / 1.32 * mult_bonsai['M_heat_HOUS']
    else:
        # Herberigs (2025): 38.43 kg CO2eq/GJ
        ghg_heat_districtheating = energy_districtheating * 38.43

# Result
ghg_heat_districtheating  # kg CO2eq per household per year
```

#### Electricity generation technology

```python
# --- Photovoltaic (PV) embodied GHG per household ---
if scenario == "Boekel":
    # Fixed GHG emissions for PV in Boekel scenario
    capacity_photovoltaic = use['solarcapacity' + year].copy()
    capacity_photovoltaic.loc[:] = 0.0  # set PV capacity to 0 (if needed downstream)
    ghg_material_photovoltaic = capacity_photovoltaic.copy()
    ghg_material_photovoltaic.loc[:] = 2940  # kg CO2eq per household (fixed total)

else:
    # Standard calculation based on ecoinvent
    capacity_photovoltaic = use['solarcapacity' + year]  # kWp PV per household
    ghg_material_photovoltaic = capacity_photovoltaic / 3 * 6313  # kg CO2eq per household

    (use['solarcapacity'+year]*use['n_hhd']/(use['n_solar'+year].fillna(2.5)))/ 3

# Result
ghg_material_photovoltaic  # kg CO2eq per household
```

### Mobility

```python
# --- Mobility GHG (pkm/vkm) ---
if scenario == "Boekel":
    # Load distance data per income group
    boekel_mob = pd.read_excel("data_input/Boekel_mobility.xlsx")  # columns: inc_lab, car_vkm, transit_pkm, cycle_pkm
    boekel_mob = boekel_mob.set_index('inc_lab')

    # Map income group (inc_lab) from `use` to distances
    inc_index = use.index.get_level_values(lab)
    car_vkm_series     = pd.Series(inc_index.map(boekel_mob['car_vkm']),     index=use.index).astype(float)
    transit_pkm_series = pd.Series(inc_index.map(boekel_mob['transit_pkm']), index=use.index).astype(float)
    cycle_pkm_series   = pd.Series(inc_index.map(boekel_mob['cycle_pkm']),   index=use.index).astype(float)

    # --- Apply emission factors ---
    # Start with base factors (average car)
    car_factor = pd.Series(0.193, index=boekel_mob.index)

    # Replace for deciles 9 and 10 (electric cars)
    for dec in ['decile 9', 'top 1% (richest)']:
        if dec in car_factor.index:
            car_factor.loc[dec] = 0.067

    # Map factors to the use index
    car_factor_series = pd.Series(inc_index.map(car_factor), index=use.index).astype(float)

    # Calculate GHG emissions (kg CO2eq per household per year)
    ghg_car_mobilityuse     = car_vkm_series     * car_factor_series
    ghg_transit_mobilityuse = transit_pkm_series * 0.020
    ghg_cycle_mobilityuse   = cycle_pkm_series   * 0.0008

else:
    # --- Original factors for non-Boekel scenarios ---
    if ghg_pkm_source == 'Milieucentraal':
        ghg_car_mobilityuse     = use['car_vkm']     * 0.193
        ghg_transit_mobilityuse = use['transit_pkm'] * 0.020
        ghg_cycle_mobilityuse   = use['cycle_pkm']   * 0.0008
    elif ghg_pkm_source == 'Milieucentraal - e-car':
        ghg_car_mobilityuse     = use['car_vkm']     * 0.067
        ghg_transit_mobilityuse = use['transit_pkm'] * 0.020
        ghg_cycle_mobilityuse   = use['cycle_pkm']   * 0.0008
    else:
        ghg_car_mobilityuse     = use['car_vkm']*0
        ghg_transit_mobilityuse = use['transit_pkm']*0
        ghg_cycle_mobilityuse   = use['cycle_pkm']*0
```
